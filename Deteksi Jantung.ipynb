```python
import numpy as np
import pandas as pd
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix, accuracy_score
```

```python
irisDataset = pd.read_csv('heart.csv', delimiter=';', header=0)
irisDataset.head()
```

```python
print(irisDataset['target'].unique())
```

```python
irisDataset["target"]=pd.factorize(irisDataset.target)[0]
print(irisDataset["target"])
```

```python
irisDataset = irisDataset.drop(labels="sex", axis=1)
print(irisDataset)
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
enc = LabelEncoder()
```

```python
irisDataset['age']=enc.fit_transform(irisDataset['age'].values)
irisDataset['cp']=enc.fit_transform(irisDataset['cp'].values)
irisDataset['trestbps']=enc.fit_transform(irisDataset['trestbps'].values)
irisDataset['chol']=enc.fit_transform(irisDataset['chol'].values)
irisDataset['fbs']=enc.fit_transform(irisDataset['fbs'].values)
irisDataset['restecg']=enc.fit_transform(irisDataset['restecg'].values)
irisDataset['thalach']=enc.fit_transform(irisDataset['thalach'].values)
irisDataset['exang']=enc.fit_transform(irisDataset['exang'].values)
irisDataset['oldpeak']=enc.fit_transform(irisDataset['oldpeak'].values)
irisDataset['ca']=enc.fit_transform(irisDataset['ca'].values)
irisDataset['thal']=enc.fit_transform(irisDataset['thal'].values)
```

```python
trs_irisDataset=irisDataset.drop(columns='target')
trs_irisDataset.head()
```

```python
lbl_irisDataset=irisDataset['target']
lbl_irisDataset.head()
```

```python
#membagi data training dan testing 0.3 data testing dan di loop sebanyak 2 kali
X_train, X_test, Y_train, Y_test = train_test_split(trs_irisDataset,lbl_irisDataset,test_size=0.2,random_state=2)
classifier=GaussianNB()
classifier.fit(X_train,Y_train)
y_pred=classifier.predict(X_test)
```

```python
print('Accuracy of Naive Bayes continue classifier on test set: {:.2f}'.format(classifier.score(X_test,Y_test)))
print(confusion_matrix(Y_test,y_pred))
print(accuracy_score(Y_test,y_pred))
hasilpengujian=classification_report(Y_test,y_pred)
print(hasilpengujian)
```

```python
irisDataset = irisDataset.to_numpy()
print(irisDataset)
```

```python
#data 80 & 20
dataTraining = np.concatenate((irisDataset[0:40, :], irisDataset[50:90, :]), axis=0)
dataTesting = np.concatenate((irisDataset[40:50, :], irisDataset[90:100, :]), axis=0)
print(dataTraining)
print()
print(dataTesting)
```

```python
inputTraining = dataTraining[:,0:4]
inputTesting = dataTesting[:,0:4]
labelTraining = dataTraining[:,4]
labelTesting = dataTesting[:,4]
print(labelTraining)
print()
len(labelTraining)
```

```python
model = GaussianNB()
```

```python
model = model.fit(inputTraining, labelTraining)
```

```python
hasilPrediksi = model.predict(inputTesting)
print("Label Sebenarnya :", labelTesting)
print("Hasil Prediksi :", hasilPrediksi)
```

```python
#hasil prediksi 80 & 20
prediksiBenar = (hasilPrediksi == labelTesting).sum()
prediksiSalah = (hasilPrediksi != labelTesting).sum()
print("Prediksi benar :", prediksiBenar, "data")
print("Prediksi salah :", prediksiSalah, "data")
print("Akurasi :", prediksiBenar/(prediksiBenar+prediksiSalah)*100, "%")
```
